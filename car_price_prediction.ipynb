```python
import numpy as np
import pandas as pd
df = pd.read_csv("CarPrice_Assignment.csv")
df.head()
```

```python
shape = df.shape
print("shape of the dataset:", shape)
columns = df.columns
print("\ncolumns of the dataset:", columns)
```

```python
describe = df.describe()
print("describe of the dataset:", describe)
```

```python
missing_values = df.isnull().sum()
print("\nmissing values in the dataset:", missing_values)

duplicate_rows = df.duplicated().sum()
print("\nduplicate rows in the dataset:", duplicate_rows)
```

```python
df = df.drop(columns=["car_ID"])
df["Company"] = (df["CarName"].str.split().str[0].str.lower())
df = df.drop(columns=["CarName"])
df.head()
```

```python
df = pd.get_dummies(
    df,
    drop_first=True,
    dtype=int
)
df.head()
```

```python
print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

df.head()
```

```python
df["price"].plot(kind="hist", bins=25, title="Distribution of Car Prices")
```

```python
correlations = df.corr(numeric_only=True)["price"]

correlations = correlations.drop("price")
correlations = correlations.sort_values(ascending=False)
print(correlations.head(10))
```

```python
X = df.drop(columns=["price"]).to_numpy(dtype=float)

y = df["price"].to_numpy(dtype=float)

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python

# Set a fixed random seed
np.random.seed(42)

# Create shuffled indices for all 205 cars
indices = np.random.permutation(len(X))

# Decide the size of the training dataset
split = int(0.8 * len(X))

# Separate the indices into training and testing groups
train_indices = indices[:split]
test_indices = indices[split:]

# Select the training data
X_train = X[train_indices]
y_train = y[train_indices]

# Select the testing data
X_test = X[test_indices]
y_test = y[test_indices]

# Display the sizes
print("Training features:", X_train.shape)
print("Training prices:", y_train.shape)
print("Testing features:", X_test.shape)
print("Testing prices:", y_test.shape)
```

```python

# Calculate the mean and standard deviation of training features
X_mean = X_train.mean(axis=0)
X_std = X_train.std(axis=0)

# Avoid division by zero
X_std[X_std == 0] = 1

# Scale training and testing features
X_train_scaled = (X_train - X_mean) / X_std
X_test_scaled = (X_test - X_mean) / X_std

# Scale the target prices using training data statistics
y_mean = y_train.mean()
y_std = y_train.std()

y_train_scaled = (y_train - y_mean) / y_std

print("Scaled training features:", X_train_scaled.shape)
print("Scaled testing features:", X_test_scaled.shape)
print("Scaled training prices:", y_train_scaled.shape)
```

```python
X_train_b = np.column_stack([
    np.ones(len(X_train_scaled)),
    X_train_scaled
])

X_test_b = np.column_stack([
    np.ones(len(X_test_scaled)),
    X_test_scaled
])

print("Training data with bias:", X_train_b.shape)
print("Testing data with bias:", X_test_b.shape)
```

```python

epochs = 2000
learning_rate = 0.001

weights = np.zeros(X_train_b.shape[1])
loss_history = []

for epoch in range(epochs):
    predictions = X_train_b @ weights
    errors = predictions - y_train_scaled

    gradients = (2 / len(y_train_scaled)) * (X_train_b.T @ errors)
    weights = weights - learning_rate * gradients

    # Calculate loss after updating the weights
    predictions = X_train_b @ weights
    errors = predictions - y_train_scaled
    loss = np.mean(errors ** 2)
    loss_history.append(loss)

print("Training completed!")
print("Initial loss:", loss_history[0])
print("Final loss:", loss_history[-1])
```

```python
pd.Series(loss_history).plot(
    title="Training Loss vs Epoch",
    xlabel="Epoch",
    ylabel="Mean Squared Error",
    figsize=(8, 5)
)
```

```python

# Predict prices for the test cars
y_pred_scaled = X_test_b @ weights

# Convert predictions back to the original price scale
y_pred = (y_pred_scaled * y_std) + y_mean

# Display actual and predicted prices
results = pd.DataFrame({
    "Actual Price": y_test,
    "Predicted Price": y_pred
})

print(results.head(10).round(2))
```

```python
y_pred_scaled = X_test_b @ weights
y_pred = y_pred_scaled * y_std + y_mean

print("First 5 scaled predictions:", y_pred_scaled[:5])
print("First 5 predicted prices:", y_pred[:5])
```

```python

# Recalculate predictions using the latest trained weights
y_pred_scaled = X_test_b @ weights
y_pred = y_pred_scaled * y_std + y_mean

# Calculate prediction errors
errors = y_test - y_pred

# 1. Mean Absolute Error (MAE)
mae = np.mean(np.abs(errors))

# 2. Mean Squared Error (MSE)
mse = np.mean(errors ** 2)

# 3. Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)

# 4. R-squared (R2)
ss_res = np.sum(errors ** 2)
ss_tot = np.sum((y_test - np.mean(y_test)) ** 2)
r2 = 1 - (ss_res / ss_tot)

# Display the metrics
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R-squared:", round(r2, 4))
```

# Car Price Prediction Using Linear Regression

## 1. Objective
The objective of this project is to build a machine learning model that predicts the price of a car based on its features, such as engine size, horsepower, fuel type, car body, and other specifications.

## 2. Dataset
The project uses the Car Price Prediction dataset. The dataset contains 205 rows of car records. The target variable is `price`, and the remaining selected columns are used as input features.

## 3. Data Preprocessing
The dataset was checked for missing values and duplicate rows. The car ID column was removed because it does not provide useful information for predicting price. The car name was processed to extract the company name. Categorical variables were converted into numerical features using one-hot encoding.

## 4. Model Development
A linear regression model was implemented from scratch using NumPy and Pandas, without using scikit-learn. The model was trained using gradient descent. The input features and target prices were scaled to help the training process.

## 5. Model Evaluation
The model was evaluated on a separate test set using Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R-squared (R²). Actual and predicted prices were also compared using a scatter plot.

## 6. Conclusion
The model learned a relationship between car specifications and car prices. The evaluation metrics and prediction graph were used to assess its performance. This project helped demonstrate data preprocessing, linear regression, gradient descent, and model evaluation using NumPy and Pandas.
